### 은행 계좌 잔액 조회 업무 흐름
1. 사용자가 은행 계좌 잔액을 조회하는 요청이 온다.
2. 사용자 요청이 어떤 서비스를 요구하는지를 분석해서 그 작업에 맞는 subagent에게 작업을 분배하는 역할을 하는 supervisor 노드가 이 작업을 bank_account_balance라는 subagent에게 보낸다.
3. 만약 사용자 요청안에 잔액 조회를 원하는 account_id가 있는지 확인해서 있다면 tool인 get_balance_by_owner(ower_id, account_id)를 갖고 있는 graph(=create_agent(model, llm, checkpointer)를 
    invoke해서 구하고 만약 account_id가 없다면 graph가 해당 owner_id는 몇개의 account를 갖고 있는지 확인해서 그 정보를 Human-in-the-loop의 Command, interrupt를 이용해서 사용자로부터 선택을 받아서
    거기에 맞게 정보를 제공한다.)

```python
from pydantic import BaseModel, Field
from typing import TypedDict, Annotated
from langgraph.graph.message import add_messages
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain.agents import create_agent
import os
from dotenv import load_dotenv

load_dotenv()

GEMINI_MODEL = os.getenv('GEMINI_MODEL')
llm = ChatGoogleGenerativeAI(model=GEMINI_MODEL)
```

```python
class Account(BaseModel):
    account_id: str = Field(description="계좌 번호")
    owner_id: str = Field(description="계좌 소유주의 아이디")
    nickname: str = Field(description="계좌의 주된 용도")
    balance: int = Field(description="계좌 잔액")

class AccountState(TypedDict):
    messages: Annotated[list, add_messages]
    owner_account: list[Account]
```

```python
@tool(parse_docstring=True)
def get_balance_by_owner(owner_id: str, account_id: str | None = None) -> str:
    """계좌 잔액을 조회한다.
    
    Args:
        owner_id: 계좌 소유주의 아이디
        account_id: 조회할 계좌 번호. 생략하면 (None) 해당 owner_id의 모든 계좌를 조회한다.
    """
    file_path = 'data/data.json'

    with open(file_path, 'r', encoding='utf-8') as f:
        data = json.load(f)

    matched = [
        account for account in data['accounts']
        if account['owner_id'] == owner_id and (account_id is None or account['account_id'] == account_id)
    ]

    if not matched:
        return json.dumps({'found': False, 'accounts': []})

    return json.dumps({
        'found': True,
        'accounts': [
            {'account_id': acc['account_id'], 'balance': acc['balance']} for acc in matched
        ]
    })
```

```python
def account_agent(state: AccountState):
    # 사용자 요청을 받는다
    request = state['messages'][-1]
    graph = create_agent(model=llm, tools=[get_balance_by_owner])
```
